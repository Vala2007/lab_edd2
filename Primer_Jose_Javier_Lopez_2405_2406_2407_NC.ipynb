#$$f(n)=b0*\left(\frac{1-\sqrt{5}}{2}\right)^n+b1*\left(\frac{1+\sqrt{5}}{2}\right)^n$$
#\begin{align}
&\texttt{Ecuaciones}\\
f(0)&=0=b0+b1\\
f(1)&=1=b0*\left(\frac{1-\sqrt{5}}{2}\right)+b1*\left(\frac{1+\sqrt{5}}{2}\right)\end{align}

```python
#Cómo hallar los bi?
import sympy as sp
from sympy import Matrix
c=Matrix([0,1])#Aquí van las condiciones iniciales
b = sp.symbols('b0:2')#Aquí cambias 4 por el número de c.i.
A = Matrix([
    [1, 1],
    [(1-sp.sqrt(5))/2,(1+sp.sqrt(5))/2]
])  # Aquí la ecuación correspondiente a cada c.i., sin incluir el valor de la c.i.
b= Matrix(b)
b = sp.solve(A*b-c,rational=True)
print(f"Los coeficientes bi son:\n{b}\n")

```
